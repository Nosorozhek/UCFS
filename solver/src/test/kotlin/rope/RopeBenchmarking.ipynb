```kotlin
%useLatestDescriptors
%use kandy
%use dataframe
```

```kotlin
fun testString(n: Int) = generateSequence { "test".asSequence() }.flatten().take(n).joinToString("")
```

```kotlin
import org.jetbrains.kotlinx.kandy.letsplot.multiplot.context.PlotBunchBuffer
import rope.BenchmarkResult

fun PlotBunchBuffer.plotResults(
    results: DataFrame<BenchmarkResult>
) {
    add(results.plot {
        x("size") {
            axis.name = "Size of string"
        }
        y("meanTime") {
            axis.name = "Time, ms"
        }
        line {
            color("name")
            type = LineType.LONGDASH
        }
        points {
            size = 3.0
            symbol = Symbol.CIRCLE_OPEN
            color("name")
        }
        errorBars {
            yMin("minTime75")
            yMax("maxTime75")
            borderLine.type = LineType.DASHED
        }
    }, 0, 0, 600, 400)
    add(results.plot {
        x("size") {
            axis.name = "Size of string"
        }
        y("meanMemoryKb") {
            axis.name = "Memory, Kb"
        }
        line {
            color("name")
            type = LineType.LONGDASH
        }
        points {
            size = 3.0
            symbol = Symbol.CIRCLE_OPEN
            color("name")
        }
        errorBars {
            yMin("minMemoryKb75")
            yMax("maxMemoryKb75")
            borderLine.type = LineType.DASHED
        }
    }, 600, 0, 600, 400)
}
```

```kotlin
import rope.ConcatRopesBenchmark
import rope.ConcatStringsBenchmark

val concatResults: DataFrame<BenchmarkResult> = (listOf(100000, 300000, 600000, 900000).map { i ->
    ConcatRopesBenchmark().run(50, 100, testString(i))
} + listOf(100000, 300000, 600000, 900000).map { i ->
    ConcatStringsBenchmark().run(50, 100, testString(i))
}).toDataFrame()
```

```kotlin
plotBunch { plotResults(concatResults as DataFrame<BenchmarkResult>) }
```

```kotlin
import rope.ConcatMultipleRopesBenchmark
import rope.ConcatMultipleStringsBenchmark

val concatMultipleResults: DataFrame<BenchmarkResult> = (listOf(1000, 3000, 6000, 9000).map { i ->
    ConcatMultipleRopesBenchmark(100).run(50, 100, testString(i))
} + listOf(1000, 3000, 6000, 9000).map { i ->
    ConcatMultipleStringsBenchmark(100).run(50, 100, testString(i))
}).toDataFrame()
```

```kotlin
plotBunch { plotResults(concatMultipleResults as DataFrame<BenchmarkResult>) }
```

```kotlin
import rope.SubRopeBenchmark
import rope.SubStringBenchmark

val subSequenceResults: DataFrame<BenchmarkResult> = (listOf(1000, 3000, 6000, 9000).map { i ->
    SubRopeBenchmark(1000).run(50, 100, testString(i))
} + listOf(1000, 3000, 6000, 9000).map { i ->
    SubStringBenchmark(1000).run(50, 100, testString(i))
}).toDataFrame()
```
